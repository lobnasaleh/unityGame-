```python
import tensorflow as tf
import numpy as np
import pandas as pd
import sentencepiece as spm
import nltk
import os
import re

# Download NLTK data
nltk.download('punkt')
```

```python
# Hyperparameters
D_MODEL = 128
NUM_HEADS = 8
NUM_LAYERS = 4
D_FF = 512
DROPOUT_RATE = 0.1
MAX_LENGTH = 20
VOCAB_SIZE = 8000
BATCH_SIZE = 64
EPOCHS = 30
SRC_PAD_IDX = 0

# File paths (Colab-specific)
DATA_PATH = '/content/ara_.txt'  # Arabic-English dataset in Colab
SPM_AR_PATH = '/content/ar_spm.model'  # SentencePiece model for Arabic
SPM_EN_PATH = '/content/en_spm.model'  # SentencePiece model for English
```

```python
def load_data(path):
    data = pd.read_csv(path, sep='\t', header=None, names=['en', 'ar'], encoding='utf-8')
    return data['en'].values, data['ar'].values

def clean_text(text):
    if not isinstance(text, str):
        return ""
    text = text.lower() if not any('\u0600' <= c <= '\u06FF' for c in text) else text
    text = re.sub(r'[^\w\s]', '', text)
    if any('\u0600' <= c <= '\u06FF' for c in text):
        text = re.sub(r'[\u0617-\u061A\u064B-\u0652]', '', text)
        text = re.sub(r'[\u0622\u0623\u0625]', '\u0627', text)
    return text
```

```python
def train_sentencepiece(data, model_prefix, vocab_size=8000):
    temp_file = f'/content/{model_prefix}.txt'
    with open(temp_file, 'w', encoding='utf-8') as f:
        for line in data:
            if line.strip():
                f.write(str(line) + '\n')
    spm.SentencePieceTrainer.train(
        input=temp_file,
        model_prefix=f'/content/{model_prefix}',
        vocab_size=vocab_size,
        character_coverage=1.0,
        model_type='bpe'
    )
    os.remove(temp_file)
```

```python
def preprocess_data():
    en_sentences, ar_sentences = load_data(DATA_PATH)
    ar_sentences = [clean_text(sent) for sent in ar_sentences]
    en_sentences = [clean_text(sent) for sent in en_sentences]

    pairs = [(ar, en) for ar, en in zip(ar_sentences, en_sentences) if ar.strip() and en.strip()]
    ar_sentences, en_sentences = zip(*pairs)

    print("Sample sentence pairs:")
    for ar, en in list(zip(ar_sentences, en_sentences))[:5]:
        print(f"Arabic: {ar}\tEnglish: {en}")

    if not os.path.exists(SPM_AR_PATH):
        train_sentencepiece(ar_sentences, 'ar_spm', VOCAB_SIZE)
    if not os.path.exists(SPM_EN_PATH):
        train_sentencepiece(en_sentences, 'en_spm', VOCAB_SIZE)

    sp_ar = spm.SentencePieceProcessor()
    sp_en = spm.SentencePieceProcessor()
    sp_ar.load(SPM_AR_PATH)
    sp_en.load(SPM_EN_PATH)

    ar_tokens = [sp_ar.encode(sent, out_type=int) for sent in ar_sentences]
    en_tokens = [[sp_en.bos_id()] + sp_en.encode(sent, out_type=int) + [sp_en.eos_id()] for sent in en_sentences]

    pairs = [(ar, en) for ar, en in zip(ar_tokens, en_tokens) if len(ar) <= MAX_LENGTH and len(en) <= MAX_LENGTH]
    ar_tokens, en_tokens = zip(*pairs)

    ar_padded = tf.keras.preprocessing.sequence.pad_sequences(ar_tokens, maxlen=MAX_LENGTH, padding='post')
    en_padded = tf.keras.preprocessing.sequence.pad_sequences(en_tokens, maxlen=MAX_LENGTH, padding='post')

    dataset = tf.data.Dataset.from_tensor_slices(((ar_padded, en_padded[:, :-1]), en_padded[:, 1:]))
    dataset = dataset.cache().shuffle(10000).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

    return dataset, sp_ar, sp_en
```

```python
def positional_encoding(max_length, d_model):
    position = np.arange(max_length)[:, np.newaxis]
    div_term = np.exp(np.arange(0, d_model, 2) * -(np.log(10000.0) / d_model))
    pe = np.zeros((max_length, d_model))
    pe[:, 0::2] = np.sin(position * div_term)
    pe[:, 1::2] = np.cos(position * div_term)
    return tf.cast(pe, dtype=tf.float32)
```

```python
class MultiHeadAttention(tf.keras.layers.Layer):
    def __init__(self, d_model, num_heads):
        super().__init__()
        self.mha = tf.keras.layers.MultiHeadAttention(num_heads=num_heads, key_dim=d_model//num_heads)
        self.layernorm = tf.keras.layers.LayerNormalization(epsilon=1e-6)

    def call(self, query, key, value, mask=None):
        attn_output = self.mha(query, key, value, attention_mask=mask)
        return self.layernorm(query + attn_output)

class FeedForward(tf.keras.layers.Layer):
    def __init__(self, d_model, d_ff):
        super().__init__()
        self.seq = tf.keras.Sequential([
            tf.keras.layers.Dense(d_ff, activation='relu'),
            tf.keras.layers.Dense(d_model)
        ])
        self.layernorm = tf.keras.layers.LayerNormalization(epsilon=1e-6)

    def call(self, x):
        return self.layernorm(x + self.seq(x))

class EncoderLayer(tf.keras.layers.Layer):
    def __init__(self, d_model, num_heads, d_ff, dropout):
        super().__init__()
        self.self_attn = MultiHeadAttention(d_model, num_heads)
        self.ffn = FeedForward(d_model, d_ff)
        self.dropout = tf.keras.layers.Dropout(dropout)

    def call(self, x, mask=None, training=False):
        x = self.self_attn(x, x, x, mask)
        x = self.dropout(x, training=training)
        x = self.ffn(x)
        return x


class DecoderLayer(tf.keras.layers.Layer):
    def __init__(self, d_model, num_heads, d_ff, dropout):
        super().__init__()
        self.self_attn = MultiHeadAttention(d_model, num_heads)
        self.cross_attn = MultiHeadAttention(d_model, num_heads)
        self.ffn = FeedForward(d_model, d_ff)
        self.dropout = tf.keras.layers.Dropout(dropout)

    def call(self, x, enc_output, look_ahead_mask=None, padding_mask=None, training=False):
        x = self.self_attn(x, x, x, look_ahead_mask)
        x = self.cross_attn(x, enc_output, enc_output, padding_mask)
        x = self.dropout(x, training=training)
        x = self.ffn(x)
        return x
```

```python
class Encoder(tf.keras.layers.Layer):
    def __init__(self, num_layers, d_model, num_heads, d_ff, input_vocab_size, dropout):
        super().__init__()
        self.d_model = d_model
        self.embedding = tf.keras.layers.Embedding(input_vocab_size, d_model)
        self.pos_encoding = positional_encoding(MAX_LENGTH, d_model)
        self.enc_layers = [EncoderLayer(d_model, num_heads, d_ff, dropout)
                          for _ in range(num_layers)]
        self.dropout = tf.keras.layers.Dropout(dropout)

    def call(self, x, mask=None, training=False):
        seq_len = tf.shape(x)[1]
        x = self.embedding(x)
        x *= tf.math.sqrt(tf.cast(self.d_model, tf.float32))
        x += self.pos_encoding[tf.newaxis, :seq_len, :]
        x = self.dropout(x, training=training)

        for layer in self.enc_layers:
            x = layer(x, mask=mask, training=training)

        return x

class Decoder(tf.keras.layers.Layer):
    def __init__(self, num_layers, d_model, num_heads, d_ff, target_vocab_size, dropout):
        super().__init__()
        self.d_model = d_model
        self.embedding = tf.keras.layers.Embedding(target_vocab_size, d_model)
        self.pos_encoding = positional_encoding(MAX_LENGTH, d_model)
        self.dec_layers = [DecoderLayer(d_model, num_heads, d_ff, dropout)
                          for _ in range(num_layers)]
        self.dropout = tf.keras.layers.Dropout(dropout)

    def call(self, x, enc_output, look_ahead_mask=None, padding_mask=None, training=False):
        seq_len = tf.shape(x)[1]
        x = self.embedding(x)
        x *= tf.math.sqrt(tf.cast(self.d_model, tf.float32))
        x += self.pos_encoding[tf.newaxis, :seq_len, :]
        x = self.dropout(x, training=training)

        for layer in self.dec_layers:
            x = layer(x, enc_output, look_ahead_mask, padding_mask, training=training)

        return x
```

```python
class Transformer(tf.keras.Model):
    def __init__(self, num_layers, d_model, num_heads, d_ff,
                 input_vocab_size, target_vocab_size, dropout=0.1):
        super().__init__()
        self.encoder = Encoder(num_layers, d_model, num_heads, d_ff,
                              input_vocab_size, dropout)
        self.decoder = Decoder(num_layers, d_model, num_heads, d_ff,
                              target_vocab_size, dropout)
        self.final_layer = tf.keras.layers.Dense(target_vocab_size)

    def call(self, inputs, training=False):
        src, trg = inputs

        src_mask = self.make_src_mask(src)  # boolean mask
        trg_mask = self.make_trg_mask(trg)  # boolean mask

        enc_output = self.encoder(src, mask=src_mask, training=training)
        dec_output = self.decoder(
            trg, enc_output,
            look_ahead_mask=trg_mask,
            padding_mask=src_mask,
            training=training
        )

        return self.final_layer(dec_output)

    def make_src_mask(self, src):
        """Create boolean padding mask (True = masked)"""
        mask = tf.math.equal(src, 0)  # boolean
        return mask[:, tf.newaxis, tf.newaxis, :]  # (batch, 1, 1, seq_len)

    def make_trg_mask(self, trg):
        """Create combined look-ahead + padding mask (boolean)"""
        seq_len = tf.shape(trg)[1]

        # 1. Look-ahead mask (prevents attending to future positions)
        look_ahead = 1 - tf.linalg.band_part(
            tf.ones((seq_len, seq_len), dtype=tf.float32),
            -1, 0  # Lower triangular including diagonal
        )
        look_ahead = tf.cast(look_ahead, tf.bool)  # Convert to boolean
        look_ahead = look_ahead[tf.newaxis, tf.newaxis, :, :]  # (1, 1, seq_len, seq_len)

        # 2. Padding mask from target input
        padding = self.make_src_mask(trg)  # (batch, 1, 1, seq_len)

        # 3. Combine using logical OR (True = masked)
        combined_mask = tf.logical_or(look_ahead, padding)
        return combined_mask
```

```python
class CustomSchedule(tf.keras.optimizers.schedules.LearningRateSchedule):
    def __init__(self, d_model, warmup_steps=4000):
        super().__init__()
        self.d_model = tf.cast(d_model, tf.float32)
        self.warmup_steps = warmup_steps

    def __call__(self, step):
        step = tf.cast(step, tf.float32)
        arg1 = tf.math.rsqrt(step)
        arg2 = step * (self.warmup_steps ** -1.5)
        return tf.math.rsqrt(self.d_model) * tf.math.minimum(arg1, arg2)
```

```python
def loss_function(real, pred):
    mask = tf.math.logical_not(tf.math.equal(real, 0))
    loss = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True,
                                                       reduction='none')(real, pred)
    mask = tf.cast(mask, dtype=loss.dtype)
    loss *= mask
    return tf.reduce_sum(loss)/tf.reduce_sum(mask)
```

```python
def train_model():
    dataset, sp_ar, sp_en = preprocess_data()
    input_vocab_size = sp_ar.get_piece_size()
    target_vocab_size = sp_en.get_piece_size()

    model = Transformer(NUM_LAYERS, D_MODEL, NUM_HEADS, D_FF,
                       input_vocab_size, target_vocab_size, DROPOUT_RATE)

    # Build model with sample input
    sample_src = tf.random.uniform((1, MAX_LENGTH), maxval=input_vocab_size, dtype=tf.int32)
    sample_trg = tf.random.uniform((1, MAX_LENGTH), maxval=target_vocab_size, dtype=tf.int32)
    model((sample_src, sample_trg))

    learning_rate = CustomSchedule(D_MODEL)
    optimizer = tf.keras.optimizers.Adam(learning_rate, beta_1=0.9, beta_2=0.98, epsilon=1e-9)

    model.compile(optimizer=optimizer, loss=loss_function, metrics=['accuracy'])

    model.fit(dataset, epochs=EPOCHS,
             callbacks=[tf.keras.callbacks.TensorBoard('logs')])

    return model, sp_ar, sp_en
```

```python
def translate(model, sp_ar, sp_en, sentence):
    """Translate Arabic text to English with dtype fixes."""
    sentence = clean_text(sentence)

    # 1. Encode and pad source text with explicit int32 dtype
    input_ids = sp_ar.encode(sentence, out_type=int)
    input_ids = tf.keras.preprocessing.sequence.pad_sequences(
        [input_ids],
        maxlen=MAX_LENGTH,
        padding='post',
        value=SRC_PAD_IDX,
        dtype='int32'  # Explicit int32 dtype
    )
    input_ids = tf.convert_to_tensor(input_ids, dtype=tf.int32)

    # 2. Precompute encoder output
    enc_output = model.encoder(input_ids, training=False)

    # 3. Initialize decoder with BOS token
    output_ids = [sp_en.bos_id()]

    # 4. Auto-regressive decoding
    for _ in range(MAX_LENGTH):
        # Prepare decoder input with int32 dtype
        decoder_input = tf.keras.preprocessing.sequence.pad_sequences(
            [output_ids],
            maxlen=MAX_LENGTH,
            padding='post',
            value=SRC_PAD_IDX,
            dtype='int32'  # Explicit int32 dtype
        )
        decoder_input = tf.convert_to_tensor(decoder_input, dtype=tf.int32)

        # Create proper boolean masks
        seq_length = decoder_input.shape[1]
        look_ahead_mask = 1 - tf.linalg.band_part(
            tf.ones((seq_length, seq_length)),
            -1, 0
        )
        look_ahead_mask = tf.cast(look_ahead_mask, tf.bool)
        padding_mask = model.make_src_mask(input_ids)

        # Run decoder
        dec_output = model.decoder(
            decoder_input,
            enc_output,
            look_ahead_mask=look_ahead_mask,
            padding_mask=padding_mask,
            training=False
        )
        predictions = model.final_layer(dec_output)

        # Get next token ID
        predicted_id = tf.argmax(predictions[0, -1], axis=-1).numpy().item()
        output_ids.append(predicted_id)

        if predicted_id == sp_en.eos_id():
            break

    # 5. Decode and return
    return sp_en.decode(output_ids[1:-1])  # Remove BOS/EOS
```

```python
from nltk.translate.bleu_score import corpus_bleu, SmoothingFunction

def evaluate_bleu(model, sp_ar, sp_en,num_samples=100):  # ✅ Correct parameter
    """Evaluate BLEU score on a subset of the data"""
    en_sentences, ar_sentences = load_data(DATA_PATH)
    smoothie = SmoothingFunction().method4

    references = []
    hypotheses = []

    valid_pairs = [(ar, en) for ar, en in zip(ar_sentences, en_sentences)
                  if ar.strip() and en.strip()][:num_samples]

    for ar, en in valid_pairs:
        try:
            translated = translate(model, sp_ar, sp_en, ar)
            if not translated.strip():
                continue

            ref_tokens = nltk.word_tokenize(en.lower())
            hyp_tokens = nltk.word_tokenize(translated.lower())

            references.append([ref_tokens])
            hypotheses.append(hyp_tokens)

        except Exception as e:
            print(f"Skipping pair due to error: {str(e)}")
            continue

    if not hypotheses:
        return 0.0

    return corpus_bleu(references, hypotheses, smoothing_function=smoothie)
```

```python
if __name__ == '__main__':
    # Train model
    model, sp_ar, sp_en = train_model()

    # Example translation
    test_sentence = "مرحبا، كيف حالك؟"  # "Hello, how are you?"
    translation = translate(model, sp_ar, sp_en, test_sentence)
    print(f"\nTest Translation:")
    print(f"Arabic: {test_sentence}")
    print(f"English: {translation}")

    # Evaluate BLEU score
    print("\nEvaluating BLEU Score...")

    print("BLEU:", evaluate_bleu(model, sp_ar, sp_en, num_samples=100))
```
